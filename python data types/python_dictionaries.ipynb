# Python Data Types
Dictionaries, lists, numbers, sets, strings and tuples.

## Dictionaries

Dictionaries store data in key-value pairs

```python
car = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}

print(car["model"])  # Accesses value by its key
```

Adding, changing and removing items:

```python
car["color"] = "red"     # add a new key
car["year"] = 1965       # change an existing value
print(car)

del car["color"]         # remove a key
print(car)
```

Using `.get()` is a safe way to read a value: it gives `None` (or a default you choose) instead of an error when the key is missing.

```python
print(car.get("model"))
print(car.get("price"))
print(car.get("price", "not listed"))
print("brand" in car)       # check if a key exists
```

Getting all the keys, values or pairs, and looping through them:

```python
print(car.keys())
print(car.values())
print(len(car))

for key, value in car.items():
    print(key, ":", value)
```

## Lists

Python List: an ordered, changeable collection of items

```python
fruits = ["apple", "banana", "cherry"]

fruits.append("orange")  # Adds an item
print(fruits)
print(fruits[0])         # Accesses the first item
```

Indexing starts at `0`. Negative numbers count from the end, and a slice `[start:stop]` gives part of the list (the stop index is not included).

```python
print(fruits[-1])      # last item
print(fruits[1:3])     # items at index 1 and 2
print(len(fruits))     # how many items
```

Changing a list:

```python
fruits.insert(1, "mango")     # add at a position
print(fruits)

fruits.remove("banana")       # remove by value
print(fruits)

last_item = fruits.pop()      # remove and return the last item
print(last_item)

fruits.sort()                 # sort alphabetically
print(fruits)
print("apple" in fruits)
```

```python
numbers = [5, 2, 9, 1]
numbers.sort()
print(numbers)
numbers.reverse()
print(numbers)

for number in numbers:
    print(number * 2)
```

## Numbers and mathematics

python numbers and mathematics

```python
# Numbers and basic math
x = 10
y = 3

print(x + y)  # Addition
print(x * y)  # Multiplication
print(x ** y) # Exponent (10 to the power of 3)
```

Python has two main number types: `int` (whole numbers) and `float` (decimals). Some more operations:

```python
x = 10
y = 3

print(x - y)             # subtraction
print(x / y)             # division always gives a float
print(x // y)            # floor division (whole number part)
print(x % y)             # remainder
print(round(x / y, 2))   # round to 2 decimal places
print(abs(-7))           # absolute value
print(max(4, 9, 2), min(4, 9, 2))
print(type(x), type(3.5))
```

## Sets

Sets are unordered and do not allow duplicate values

```python
my_set = {"apple", "banana", "cherry", "apple"}

print(my_set)  # Notice duplicate "apple" is automatically removed
```

Sets can be changed with `.add()` and `.remove()`. Because sets are unordered, the items may print in a different order each time you run the code.

```python
my_set.add("orange")
my_set.add("apple")        # already there, so nothing changes
my_set.remove("banana")

print(my_set)
print("cherry" in my_set)
print(len(my_set))
```

Sets are handy for comparing groups of items and for removing duplicates from a list.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)    # union: everything in a or b
print(a & b)    # intersection: only items in both
print(a - b)    # difference: in a but not in b

numbers_with_duplicates = [1, 2, 2, 3, 3, 3]
print(list(set(numbers_with_duplicates)))
```

## Strings

Strings are text values

```python
name = "victor"

print(name.upper())  # Makes it uppercase
print(len(name))     # Counts the letters
```

Useful string methods. Remember that strings cannot be changed, so these methods give back a **new** string.

```python
name = "victor muchai"

print(name.title())                    # Capitalise each word
print(name.lower())
print(name.replace("victor", "john"))
print(name.split())                    # split into a list of words
print("  hello  ".strip())              # remove spaces at the start and end
print(name.startswith("vic"))
print("muchai" in name)
```

Strings are indexed like lists: the first character is at index `0`.

```python
print(name[0])        # first character
print(name[-1])       # last character
print(name[0:6])      # characters at index 0 to 5

age = 22
print(f"{name.title()} is {age} years old")
```

## Tuples

Python Tuple: an ordered collection that cannot be changed

```python
# Tuples are ordered but unchangeable (immutable)
my_tuple = ("apple", "banana", "cherry")

print(my_tuple)
print(my_tuple[1])  # Accesses the second item
```

```python
print(len(my_tuple))
print(my_tuple.index("cherry"))     # position of an item
print(my_tuple.count("apple"))      # how many times an item appears

first, second, third = my_tuple     # unpacking
print(first, second, third)

try:
    my_tuple[0] = "mango"
except TypeError:
    print("Tuples cannot be changed!")

single_item = ("apple",)            # a one-item tuple needs a trailing comma
print(type(single_item))
```

## Summary

| Type | Example | Ordered? | Changeable? | Duplicates? |
|------|---------|----------|-------------|-------------|
| list | `[1, 2, 3]` | yes | yes | yes |
| tuple | `(1, 2, 3)` | yes | no | yes |
| set | `{1, 2, 3}` | no | yes (add/remove) | no |
| dictionary | `{"a": 1}` | yes (by insertion) | yes | keys must be unique |
