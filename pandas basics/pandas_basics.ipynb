# Pandas Basics
Pandas is a library for working with tables of data. Install it once with `pip install pandas`, then import it. By convention we shorten the name to `pd`.

```python
import pandas as pd
```

## Series
A **Series** is a single column of data with an index (labels).

```python
ages = pd.Series([20, 21, 19, 18], index=["Tom", "nick", "krish", "victor"])

print(ages)
print(ages["nick"])
print(ages.mean())
```

## DataFrame
A **DataFrame** is a whole table made of rows and columns. You can create one from a dictionary where each key is a column name.

used pandas to create data,add newcolumn and add new row

```python
#importing libraryating
import pandas as pd

#declaring variable
data={'Name':['Tom','nick','krish','victor'],
      'Age':[20,21,19,18]}
df=pd.DataFrame(data,index=['1st born','2nd born','3rd born','lastborn'])

# add a new column
df['job']=['actor','artist','chef','engineer']

# add a new row
new_rows=pd.DataFrame([{'Name':'lucy','Age':19,'job':'accountant'},{'Name':'james','Age':49,'job':'Manageer'}],index=['sister','father'])
df=pd.concat([df,new_rows])
print(df)
```

## Looking at your data

```python
print(df.head(3))              # first 3 rows
print(df.shape)               # (rows, columns)
print(list(df.columns))       # column names
print(df.index.tolist())      # row labels
print(df.dtypes)              # data type of each column
```

```python
print(df["Age"].describe())   # quick statistics for a column
```

## Selecting data

Selecting columns:

```python
print(df["Name"])             # one column (a Series)
print(df[["Name", "job"]])    # several columns (a DataFrame)
```

Selecting rows:
- `.loc[]` uses the row **labels**
- `.iloc[]` uses the row **positions** (starting from 0)

```python
print(df.loc["father"])
print(df.loc["sister", "job"])
print(df.iloc[0])
print(df.iloc[0:2])
```

## Filtering rows
Put a condition inside `df[...]` to keep only the rows where it is `True`. To combine conditions use `&` (and) / `|` (or) and put each condition in brackets.

```python
print(df[df["Age"] > 19])
print(df[df["job"] == "chef"])
print(df[(df["Age"] > 18) & (df["Age"] < 30)])
```

## Simple statistics

```python
print(df["Age"].mean())
print(df["Age"].max())
print(df["Age"].min())
print(df["Age"].idxmax())     # the row label with the largest age
```

## Changing data

```python
df.loc["lastborn", "job"] = "doctor"      # change one value
df["Age_next_year"] = df["Age"] + 1       # new column calculated from another

print(df)
```

## Sorting

```python
print(df.sort_values("Age"))
print(df.sort_values("Age", ascending=False))
```

## Removing rows and columns

```python
df = df.drop("father")                    # remove a row by its label
df = df.drop(columns=["Age_next_year"])   # remove a column

print(df)
```

## Saving to and reading from a CSV file

```python
df.to_csv("family.csv")                              # save
df_loaded = pd.read_csv("family.csv", index_col=0)   # read it back

print(df_loaded)
```
