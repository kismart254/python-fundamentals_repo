used pandas to create data,add newcolumn and add new row

```python
#importing libraryating
import pandas as pd

#declaring variable
data={'Name':['Tom','nick','krish','victor'],
      'Age':[20,21,19,18]}
df=pd.DataFrame(data,index=['1st born','2nd born','3rd born','lastborn'])

# add a new column
df['job']=['actor','artist','chef','engineer']

# add a new row
new_rows=pd.DataFrame([{'Name':'lucy','Age':19,'job':'accountant'},{'Name':'james','Age':49,'job':'Manageer'}],index=['sister','father'])
df=pd.concat([df,new_rows])
print(df)
```
