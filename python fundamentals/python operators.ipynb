Operators are special symbols in Python that carry out arithmetic or logical computation. The value that the operator operates on is called the operand

Arithmetic operators
addition(+)      Adds values together
substraction(-)  Subtracts the right operand from the left
multlplication(*) Multiplies values
modulous(%)       Returns the remainder of a division
division(/)       Divides left operand by the right (returns a decimal/float)
exponetial(**)    Raises the left number to the power of the right
floor division(//) Divides and rounds down to the nearest whole integer


```python
x = 15
y = 4
print('x + y = ',x+y)
print('x - y = ',x-y)
print('x * y = ',x*y)
print('x / y = ',x/y)
print('x // y = ',x//y)
print('x ** y = ',x**y)
```

Comparison operators
Comparison operators are used to compare values. It either returns True or False according to the condition.

```python
x = 10
y = 12
print('x > y  is',x>y)
print('x < y  is',x<y)
print('x == y is',x==y)
print('x != y is',x!=y)
print('x >= y is',x>=y)
print('x <= y is',x<=y)
```
