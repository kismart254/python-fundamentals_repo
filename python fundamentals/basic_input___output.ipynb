# Basic Input & Output, Operators and Variables

In Python, we use the input() function to enter input and the print() function to print output

```python
name=input('Enter your name')
Year_of_birth=input('enter year of birth')
current_year=input('Enter current year')
age=int(current_year)-int(Year_of_birth)
print(f"my name is {name}, age {age}")
```

### `input()` always gives back text
Even if the user types a number, `input()` returns it as a **string**. To do maths with it, convert it using `int()` or `float()`.

```python
age_text = "22"            # imagine the user typed 22
print(type(age_text))

age_number = int(age_text)
print(type(age_number))
print(age_number + 1)
```

Another example: the area of a rectangle (the user types `5` and `3.5`).

```python
length = float(input("Enter length: "))
width = float(input("Enter width: "))
area = length * width
print(f"The area is {area}")
```

### Different ways to use `print()`

```python
name = "Victor"
age = 22

print("Name:", name, "Age:", age)                  # commas
print("Name: " + name)                             # + joins strings (both sides must be strings)
print(f"{name} is {age} years old")                # f-string (recommended)
print("{} is {} years old".format(name, age))      # .format()
print("Line 1\nLine 2")                            # \n = new line
print("Column 1\tColumn 2")                        # \t = tab
```

### Arithmetic operators

| Operator | Name | What it does |
|----------|------|--------------|
| `+` | addition | adds values together |
| `-` | subtraction | subtracts the right operand from the left |
| `*` | multiplication | multiplies values |
| `/` | division | divides left by right (returns a decimal/float) |
| `//` | floor division | divides and rounds down to the nearest whole number |
| `%` | modulus | returns the remainder of a division |
| `**` | exponent | raises the left number to the power of the right |

```python
x = 15
y = 4
print('x + y = ',x+y)
print('x - y = ',x-y)
print('x * y = ',x*y)
print('x / y = ',x/y)
print('x // y = ',x//y)
print('x ** y = ',x**y)
```

The modulus operator `%` gives the remainder. For example `15 % 4` is `3`, because 4 goes into 15 three times with 3 left over.

```python
x = 15
y = 4
print('x % y = ', x % y)
```

### Order of operations
Python follows the same rules as maths: brackets first, then exponents, then multiplication/division, then addition/subtraction.

```python
print(2 + 3 * 4)       # multiplication first -> 14
print((2 + 3) * 4)     # brackets first -> 20
print(2 ** 3 + 1)      # exponent first -> 9
```

### Assignment operators
These are shortcuts for updating a variable.

```python
score = 10
score += 5     # same as score = score + 5
print(score)

score -= 3     # same as score = score - 3
print(score)

score *= 2     # same as score = score * 2
print(score)

score /= 4     # same as score = score / 4
print(score)
```

python inplicit conversion_automatically converts one data type to another 

```python
integer_number = 5
float_number = 10.25

new_number = integer_number + float_number

# display new value and resulting data type
print("Value:",new_number)
print("Data Type:",type(new_number))
```

In Explicit Type Conversion, users convert the data type of an object to required data type.

We use the built-in functions like int(), float(), str(), etc to perform explicit type conversion.

```python
num_string = '10'
num_integer = 6

print("Data type of num_string before conversion:",type(num_string))

# explicit type conversion
num_string = int(num_string)

print("Data type of num_string after conversion:",type(num_string))

num_sum = num_integer + num_string

print("Sum:",num_sum)
print("Data type of num_sum:",type(num_sum))
```

More conversion examples:

```python
print(int("42") + 8)           # string -> int
print(float("3.5") + 1)        # string -> float
print(str(100) + " apples")    # int -> string
print(int(9.99))               # float -> int cuts off the decimal (no rounding)
print(round(9.99))             # round() rounds to the nearest whole number
print(bool(0), bool(5), bool(""), bool("hi"))   # 0 and empty text are False
```

If a value can't be converted, Python raises a `ValueError`. We can catch it with `try` / `except` so the program doesn't crash.

```python
try:
    number = int("hello")
except ValueError:
    print("'hello' cannot be turned into an integer")
```

python variables and literals

```python
a = 5
b = 3.2
c = 'muchai'

print (a)  # prints 5
print (b)  # prints 3.2
print (c)  # print muchai 
```

A **variable** is a name that stores a value. A **literal** is the value itself (like `5`, `3.2` or `'muchai'`). More things you can do with variables:

```python
x, y, z = 1, 2, 3          # assign several variables at once
print(x, y, z)

a = b = 0                  # give the same value to several variables
print(a, b)

x, y = y, x                # swap two values
print(x, y)

PI = 3.14159               # CAPITAL_LETTERS is a convention for values that should not change
print(PI)

big_number = 1_000_000     # underscores make big numbers easier to read
print(big_number)

paragraph = """This is
a multi-line
string"""
print(paragraph)
```

changing the valuae of a variable in python

```python
a = 5
b = 3.2
c = 'muchai'

#new value to variable c
c = 'victor'

print (a)  # prints 5
print (b)  # prints 3.2
print (c)  # print muchai 
```

Python lets a variable change type when you give it a new value:

```python
value = 5
print(type(value))

value = "five"
print(type(value))
```
