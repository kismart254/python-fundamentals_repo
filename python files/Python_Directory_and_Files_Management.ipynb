# Python Directory and Files Management
The `os` module lets Python work with folders and files, and the `csv` module helps read and write spreadsheet-style files.

## Working with directories

```python
import os

# Get current working directory
current_dir = os.getcwd()
print(f"Current Directory: {current_dir}")

# Create a new directory
os.makedirs("test_folder", exist_ok=True)

# List files and directories in the current folder
print(os.listdir("."))
```

Checking whether a path exists, and building a file path safely with `os.path.join()`:

```python
print(os.path.exists("test_folder"))
print(os.path.isdir("test_folder"))     # is it a folder?
print(os.path.isfile("test_folder"))    # is it a file?

path = os.path.join("test_folder", "notes.txt")   # works on Windows, Mac and Linux
print(path)
```

## Text files
Use `open(filename, mode)` to work with a file. The most common modes are:

| Mode | Meaning |
|------|---------|
| `"r"` | read (default). Error if the file does not exist |
| `"w"` | write. Creates the file, or **erases** it if it already exists |
| `"a"` | append. Adds to the end of the file |

Using `with open(...)` is the best way because Python closes the file for you automatically.

Writing to a text file:

```python
with open(path, "w") as file:
    file.write("Hello, this is my first file.\n")
    file.write("Python makes files easy.\n")

print("File written")
```

Reading the whole file:

```python
with open(path, "r") as file:
    content = file.read()

print(content)
```

Reading line by line, or getting all lines as a list:

```python
with open(path) as file:
    for line in file:
        print(line.strip())     # strip() removes the extra new line

with open(path) as file:
    lines = file.readlines()

print(lines)
```

Appending adds new text to the end without erasing what is already there:

```python
with open(path, "a") as file:
    file.write("This line was appended.\n")

with open(path) as file:
    print(file.read())
```

### Renaming and deleting files

```python
new_path = os.path.join("test_folder", "renamed_notes.txt")

os.rename(path, new_path)
print(os.listdir("test_folder"))

os.remove(new_path)
print(os.listdir("test_folder"))
```

### Handling a missing file
Trying to open a file that does not exist raises a `FileNotFoundError`. We can handle it with `try` / `except`.

```python
try:
    with open("missing.txt") as file:
        print(file.read())
except FileNotFoundError:
    print("That file does not exist.")
```

## CSV files
CSV (comma-separated values) files store table data as plain text, one row per line.

Writing data to a CSV file

```python
import csv
data = [
    ["Name", "Age", "City"],
    ["Victor", "20", "Nairobi"],
    ["Alice", "25", "Mombasa"]
]

with open("people.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file written successfully!")
```

reading csv files

```python
import csv

# Reading data from the CSV file
with open("people.csv", mode="r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

writing csv file

```python
import csv

# Appending a new row to an existing CSV file
new_row = ["Bob", "22", "Kisumu"]

with open("people.csv", mode="a", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(new_row)
```

Reading the file again shows the new row:

```python
import csv

with open("people.csv", mode="r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

`DictReader` uses the first row as column names, so you can get values by name instead of by position:

```python
with open("people.csv", mode="r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row["Name"], "lives in", row["City"])
```

`DictWriter` writes a list of dictionaries:

```python
students = [
    {"Name": "Tom", "Score": 80},
    {"Name": "Lucy", "Score": 92}
]

with open("scores.csv", mode="w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=["Name", "Score"])
    writer.writeheader()
    writer.writerows(students)

with open("scores.csv") as file:
    print(file.read())
```

Pandas can read a CSV file in one line (see the pandas basics notebook):

```python
import pandas as pd

print(pd.read_csv("people.csv"))
```

Finally, clean up the practice file we no longer need:

```python
os.remove("scores.csv")
print(os.path.exists("scores.csv"))
```
