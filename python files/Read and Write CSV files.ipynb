Writing data to a CSV file

```python
import csv
data = [
    ["Name", "Age", "City"],
    ["Victor", "20", "Nairobi"],
    ["Alice", "25", "Mombasa"]
]

with open("people.csv", mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(data)

print("CSV file written successfully!")
```
