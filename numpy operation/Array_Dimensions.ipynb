# NumPy Array Basics
NumPy is a Python library for working with numbers quickly. Its main object is the **array** (`ndarray`). Install it once with `pip install numpy`, then import it. By convention we shorten the name to `np`.

```python
import numpy as np
```

### Lists vs arrays
A NumPy array looks like a list, but it holds items of **one type** and does maths on **every element at once**. With a list, `* 2` repeats the list instead.

```python
numbers_list = [1, 2, 3]
numbers_array = np.array([1, 2, 3])

print(numbers_list * 2)     # repeats the list
print(numbers_array * 2)    # multiplies each element
```

## Creating arrays

```python
import numpy as np

# Creating a 1D array from a list
arr = np.array([1, 2, 3, 4, 5])
print("1D Array:", arr)
print("Array Type:", type(arr))
```

Useful facts about an array: its data type (`dtype`), its `shape` and its `size` (total number of items).

```python
print("Data type:", arr.dtype)
print("Shape:", arr.shape)       # (5,) means 5 items in one dimension
print("Size:", arr.size)
```

NumPy can also create arrays for you:

```python
print(np.zeros(4))              # four 0s
print(np.ones(3))               # three 1s
print(np.full(3, 7))            # three 7s
print(np.arange(0, 10, 2))      # start, stop (not included), step
print(np.linspace(0, 1, 5))     # 5 evenly spaced numbers from 0 to 1
```

## Array dimensions
The number of dimensions (`ndim`) tells you how many directions the data has:

| Dimensions | Name | Example |
|------------|------|---------|
| 0-D | scalar (a single value) | `42` |
| 1-D | vector (a row of values) | `[1, 2, 3]` |
| 2-D | matrix (rows and columns) | `[[1, 2, 3], [4, 5, 6]]` |
| 3-D | a stack of matrices | `[[[1, 2], [3, 4]], [[5, 6], [7, 8]]]` |

```python
import numpy as np

# 0-D Array (Scalar)
arr0 = np.array(42)

# 1-D Array
arr1 = np.array([1, 2, 3])

# 2-D Array (Matrix format with rows and columns)
arr2 = np.array([[1, 2, 3], [4, 5, 6]])

print("2D Array:\n", arr2)
print("Number of dimensions (ndim):", arr2.ndim)
```

`ndim` and `shape` for each array we created above. For a 2-D array, the shape is `(rows, columns)`.

```python
print("0-D:", arr0.ndim, arr0.shape)
print("1-D:", arr1.ndim, arr1.shape)
print("2-D:", arr2.ndim, arr2.shape)

arr3 = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("3-D:", arr3.ndim, arr3.shape)
```

`reshape()` changes the shape of an array without changing its data. The new shape must hold the same number of items.

```python
numbers = np.arange(1, 7)
print(numbers)

print(numbers.reshape(2, 3))     # 2 rows, 3 columns
print(numbers.reshape(3, 2))     # 3 rows, 2 columns
```

## Indexing and slicing
- Indexing picks one item. The first item is at index `0`.
- Slicing `[start:stop]` picks a range. The `stop` index is **not** included.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

# Accessing elements
print("First element:", arr[0])

# Slicing (from index 1 to 4, stopping before 4)
print("Sliced array (index 1 to 3):", arr[1:4])
```

More ways to index and slice a 1-D array:

```python
print(arr[-1])        # last element
print(arr[:3])        # from the start up to index 2
print(arr[2:])        # from index 2 to the end
print(arr[::2])       # every second element

arr[0] = 99           # change a value
print(arr)
```

For a 2-D array use `[row, column]`:

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])

print(matrix[0, 1])      # row 0, column 1
print(matrix[1, 2])      # row 1, column 2
print(matrix[0])         # the whole first row
print(matrix[:, 1])      # the whole second column
```

A condition inside the brackets keeps only the items where it is `True`. This is called **boolean indexing**.

```python
scores = np.array([35, 80, 62, 91, 48])

print(scores > 50)           # True/False for each item
print(scores[scores > 50])   # only the scores above 50
```

## Array operations
Math on arrays works **element by element**.

```python
import numpy as np

arr1 = np.array([1, 2, 3])
arr2 = np.array([10, 20, 30])

# Element-wise addition
print("Addition:", arr1 + arr2)

# Multiplication
print("Multiplication:", arr1 * 3)

# Mathematical functions
print("Square root of elements:", np.sqrt(arr2))
```

The other operators work the same way. Arrays of the same shape are combined item by item.

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Subtraction:", a - b)
print("Division:", a / b)
print("Power:", b ** 2)
print("Comparison:", a > 15)
```

Summary functions turn many values into one:

```python
data = np.array([4, 8, 15, 16, 23, 42])

print("Sum:", data.sum())
print("Mean:", data.mean())
print("Max:", data.max())
print("Min:", data.min())
print("Position of the max:", data.argmax())
```

On a 2-D array you can choose a direction with `axis`: `axis=0` goes down the columns and `axis=1` goes across the rows.

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])

print(matrix.sum())            # everything
print(matrix.sum(axis=0))      # sum of each column
print(matrix.sum(axis=1))      # sum of each row
```
