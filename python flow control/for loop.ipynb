In Python, a for statement is used to loop through any iterable object, including sequences such as lists, strings and dictionaries.

```python
print("--- Example 1: Looping through a String ---")
name = "Victor"
for letter in name:
    print(f"Current letter: {letter}")

print("\n--- Example 2: Looping through a range of numbers ---")
# range(1, 6) starts at 1 and stops right before 6
for number in range(1, 6):
    print(f"Count: {number}")
    
```
