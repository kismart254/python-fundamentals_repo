# Booleans, Conditions and Loops

Boolean is a data type that represents one of two values: True or False
A Boolean Expression is any expression or statement that evaluates to either True or False, typically built using comparison operators.

```python

is_logged_in = True
has_permission = False

print(type(is_logged_in)) 

# Evaluating boolean expressions
x = 15
y = 20

print(x < y)  
print(x == y)  


age = 20
is_adult = age >= 18
print(f"Is the user an adult? {is_adult}")
```

### Comparison operators
Comparison operators compare two values and give back `True` or `False`.

| Operator | Meaning |
|----------|---------|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

Remember: `=` stores a value in a variable, while `==` compares two values.

```python
a = 10
b = 3

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= 10)
print(a <= 5)
```

### Logical operators
Logical operators combine Boolean expressions.

- `and` is `True` only if **both** sides are True
- `or` is `True` if **at least one** side is True
- `not` flips True to False and False to True

```python
age = 20
has_id = True

print(age >= 18 and has_id)    # True and True
print(age < 18 or has_id)      # False or True
print(not has_id)              # not True
```

### Truthy and falsy values
Every value in Python can be treated as `True` or `False`. Zero, empty text, empty lists and `None` are **falsy**. Almost everything else is **truthy**.

```python
print(bool(0), bool(""), bool([]), bool(None))
print(bool(1), bool("hello"), bool([1, 2]))
```

The `in` operator checks whether something is inside a string or a list. It also gives back a Boolean.

```python
print("a" in "banana")
print(3 in [1, 2, 3])
print("z" in "banana")
```

## `if` statements

In Python, the if statement allows us to run or skip a block of code based on a specified condition. This enables us to create programs that respond differently depending on the situation.

```python
age = int(input("Enter your age: "))
print('enterd age:',age)

# Check if age is 18 or more
if age >= 18:
    print("Grant access to the website.")

print("Program complete.")
```

### `if` / `elif` / `else`
- `elif` (short for "else if") checks another condition if the first one was False.
- `else` runs when none of the conditions were True.

Python checks from the top and runs only the **first** block that is True.

```python
score = 72

if score >= 70:
    print("Grade: A")
elif score >= 60:
    print("Grade: B")
elif score >= 50:
    print("Grade: C")
else:
    print("Grade: F")
```

```python
number = 7

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
```

## `for` loops

In Python, a for statement is used to loop through any iterable object, including sequences such as lists, strings and dictionaries.

```python
print("--- Example 1: Looping through a String ---")
name = "Victor"
for letter in name:
    print(f"Current letter: {letter}")

print("\n--- Example 2: Looping through a range of numbers ---")
# range(1, 6) starts at 1 and stops right before 6
for number in range(1, 6):
    print(f"Count: {number}")
    
```

Looping through a list, and using `range()` with a step:

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)

print("---")
for number in range(0, 10, 2):    # start, stop (not included), step
    print(number)
```

A loop can build up a result, for example adding numbers together. Looping through a dictionary with `.items()` gives each key and value.

```python
total = 0
for n in range(1, 6):
    total += n
print("Total:", total)

print("---")
car = {"brand": "Ford", "model": "Mustang", "year": 1964}
for key, value in car.items():
    print(key, "->", value)
```

## `while` loops

In Python, a while loop is used to run a block of code as long as the specified condition is true.

```python
# Simple while loop
i = 1

while i < 6:
    print(i)
    i += 1
```

A countdown using `while`:

```python
count = 3

while count > 0:
    print(count)
    count -= 1

print("Liftoff!")
```

**Be careful:** if the condition never becomes `False`, the loop runs forever (an *infinite loop*). Always make sure something inside the loop changes the condition, or use `break` to leave it.

```python
attempts = 0

while True:
    attempts += 1
    print("Attempt", attempts)
    if attempts == 3:
        break
```

## `break`, `continue` and `pass`

The break and continue statements let us exit loops early or skip iterations as needed, giving us more ways to control loop execution.

```python
# Break example (stops when x is 3)
print("--- Break ---")
for x in range(6):
    if x == 3:
        break
    print(x)

# Continue example (skips when x is 3)
print("\n--- Continue ---")
for x in range(6):
    if x == 3:
        continue
    print(x)
```

`break` is useful when searching: stop as soon as the item is found.

```python
names = ["Tom", "nick", "krish", "victor"]

for name in names:
    if name == "krish":
        print("Found krish!")
        break
    print("Checking", name)
```

`continue` is useful for skipping values you don't want, for example printing only odd numbers.

```python
for n in range(1, 8):
    if n % 2 == 0:
        continue
    print(n)
```

In Python, pass is a null statement and it does nothing when executed. Despite not doing anything, pass does have a purpose.

As you know, Python uses indentation to define a block of code (such as the body of if, for, while etc.). These blocks cannot be empty. 

```python
# Pass statement (does nothing when x is 2)
for x in [1, 2, 3]:
    if x == 2:
        pass
    print(x)
```

`pass` is often used as a placeholder for code you will write later:

```python
def coming_soon():
    pass    # TODO: write this later

coming_soon()
print("No error, pass just does nothing")
```
