used numpy and matplot libraries to plot a cosine and sine graph

```python
#importing libraries
import numpy as np
import matplotlib.pyplot as plt

#declaring variables
X=np.arange(0,2*np.pi,0.05)
y=np.cos(X)
b=np.arange(0,2*np.pi,0.05)
z=np.sin(X)

#plotting first graph
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.title('cosinewave graph',color='b')
plt.xlabel('independent_variable')
plt.ylabel('dependent_variable')
plt.grid(True,linestyle='--',alpha=0.5)
plt.tight_layout()
plt.legend()
plt.plot(X,y,color='b',linewidth=2)
plt.show

#plotting 2nd graph
plt.figure(figsize=(8,4))
plt.subplot(1,2,2)
plt.title('sinewave graph',color='g')
plt.xlabel('independent_variable')
plt.ylabel('dependent_variable')
plt.grid(True,linestyle='--',alpha=0.5)
plt.tight_layout()
plt.legend()
plt.plot(b,z,color='g',linewidth=2)
plt.show
```
