# Sine and Cosine Waves with NumPy and Matplotlib
In this notebook I use **NumPy** to calculate sine and cosine values and **Matplotlib** to draw them as graphs. Both waves repeat every `2π` radians, so we will plot one full cycle from `0` to `2π`.

### Step 1: Create the x values
`np.arange(start, stop, step)` makes an array of evenly spaced numbers. `np.pi` is the value of π (about 3.14159).

```python
import numpy as np

print(np.pi)

x = np.arange(0, 2 * np.pi, 0.05)
print(x[:5])        # the first 5 values
print(len(x))       # how many points we have
```

### Step 2: Calculate sine and cosine
`np.sin()` and `np.cos()` work on a whole array at once. Here are their values at the key points of the cycle:

```python
angles = np.array([0, np.pi / 2, np.pi, 3 * np.pi / 2, 2 * np.pi])

print("angle   :", np.round(angles, 2))
print("sin     :", np.round(np.sin(angles), 2) + 0)
print("cos     :", np.round(np.cos(angles), 2) + 0)
```

### Step 3: Plot one wave
`plt.plot(x, y)` draws a line. Then we add a title and axis labels and call `plt.show()` to display it.

```python
import matplotlib.pyplot as plt

plt.plot(x, np.sin(x))
plt.title("Sine wave")
plt.xlabel("x (radians)")
plt.ylabel("sin(x)")
plt.show()
```

### My sine and cosine graphs

```python
#importing libraries
import numpy as np
import matplotlib.pyplot as plt

#declaring variables
X=np.arange(0,2*np.pi,0.05)
y=np.cos(X)
b=np.arange(0,2*np.pi,0.05)
z=np.sin(X)

#plotting first graph
plt.figure(figsize=(8,4))
plt.subplot(1,2,1)
plt.title('cosinewave graph',color='b')
plt.xlabel('independent_variable')
plt.ylabel('dependent_variable')
plt.grid(True,linestyle='--',alpha=0.5)
plt.tight_layout()
plt.legend()
plt.plot(X,y,color='b',linewidth=2)
plt.show

#plotting 2nd graph
plt.figure(figsize=(8,4))
plt.subplot(1,2,2)
plt.title('sinewave graph',color='g')
plt.xlabel('independent_variable')
plt.ylabel('dependent_variable')
plt.grid(True,linestyle='--',alpha=0.5)
plt.tight_layout()
plt.legend()
plt.plot(b,z,color='g',linewidth=2)
plt.show



```

#### Notes on the code above
The graphs work, but there are a few things to improve:

1. `plt.show` needs brackets: `plt.show()`. Without them the function is not actually called.
2. `plt.legend()` shows labels, so each line needs a `label=...`. Without one, Matplotlib gives the *"No artists with labels found"* warning.
3. Each `plt.figure()` creates a **new** figure. Using `plt.subplot(1, 2, 2)` in a new figure leaves the left half empty. To put both graphs side by side, create **one** figure and use `plt.subplot(1, 2, 1)` and `plt.subplot(1, 2, 2)` inside it.

`plt.subplot(rows, columns, position)` splits the figure into a grid.

```python
x = np.arange(0, 2 * np.pi, 0.05)
y_cos = np.cos(x)
y_sin = np.sin(x)

plt.figure(figsize=(8, 4))             # one figure for both graphs

plt.subplot(1, 2, 1)                   # 1 row, 2 columns, first graph
plt.plot(x, y_cos, color="b", linewidth=2, label="cos(x)")
plt.title("cosine wave graph", color="b")
plt.xlabel("independent_variable")
plt.ylabel("dependent_variable")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

plt.subplot(1, 2, 2)                   # second graph
plt.plot(x, y_sin, color="g", linewidth=2, label="sin(x)")
plt.title("sine wave graph", color="g")
plt.xlabel("independent_variable")
plt.ylabel("dependent_variable")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

plt.tight_layout()                     # stops the graphs overlapping
plt.show()
```

Analysis of Sine and Cosine Waves
In the plots above, we observe two fundamental trigonometric wave patterns over a complete cycle from zero to two pi radians.

Sine Wave
The first subplot shows the sine wave.

It starts at zero, smoothly rises to its maximum peak of one, drops back down through zero, reaches its lowest point at negative one, and finally returns back to zero.

This wave illustrates a balanced oscillation around the horizontal axis.

Cosine Wave
The second subplot shows the cosine wave.

It starts at its highest value of one, smoothly decreases through zero down to its lowest point at negative one, and then curves back up through zero to end at its peak of one.

It follows the exact same shape and speed as the sine wave, but it starts at a different point in its cycle.

Key Observation
The two waves are identical in shape and amplitude. The cosine wave simply leads the sine wave, meaning it reaches each peak and trough earlier in the cycle.

### Both waves on one graph
Calling `plt.plot()` twice before `plt.show()` draws both lines on the same axes, which makes the difference between them easy to see.

```python
plt.figure(figsize=(8, 4))
plt.plot(x, y_sin, color="g", label="sin(x)")
plt.plot(x, y_cos, color="b", linestyle="--", label="cos(x)")
plt.title("Sine and cosine together")
plt.xlabel("x (radians)")
plt.ylabel("value")
plt.axhline(0, color="black", linewidth=0.8)    # horizontal line at y = 0
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```

### Cosine is a shifted sine
Shifting the sine wave left by `π/2` (a quarter of a cycle) gives the cosine wave: `cos(x) = sin(x + π/2)`. We can check this with NumPy. `np.allclose()` tells us whether two arrays are equal (ignoring tiny rounding differences).

```python
print(np.allclose(np.cos(x), np.sin(x + np.pi / 2)))
```

### Changing the amplitude and frequency
- **Amplitude** is how tall the wave is. Multiplying by a number changes it: `2 * np.sin(x)`.
- **Frequency** is how many cycles fit in the same space. Multiplying `x` changes it: `np.sin(2 * x)`.

```python
plt.figure(figsize=(8, 4))
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, 2 * np.sin(x), label="2 * sin(x)  (taller)")
plt.plot(x, np.sin(2 * x), label="sin(2x)  (faster)")
plt.title("Amplitude and frequency")
plt.xlabel("x (radians)")
plt.ylabel("value")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```
