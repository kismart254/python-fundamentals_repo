### More on `print()`
`print()` shows output on the screen. You can print text (inside quotes), numbers, or several things at once separated by commas.

```python
print("Hello, World!")
print(2026)
print("Python", "is", "fun")          # commas put a space between items
print("2026", "10", "14", sep="-")    # sep changes the separator
print("Loading", end="...")           # end changes what is printed at the end of the line
print("done")
```

### Variable naming rules
- A name must start with a letter or an underscore (`_`), not a number.
- It can only contain letters, numbers and underscores (no spaces or hyphens).
- Names are **case-sensitive**: `age` and `Age` are different variables.
- A name cannot be a Python keyword such as `if`, `for` or `True`.
- By convention we use `snake_case` (lowercase words joined by underscores).

> **Tip:** In the earlier cell I named a variable `sum`, but `sum` is also a built-in Python function. Naming a variable the same as a built-in hides the built-in, so `total` would be a safer name.

```python
first_name = "victor"    # good: snake_case
age = 22
Age = 30                 # a different variable from `age`
print(age, Age)

# These would cause errors, so they are commented out:
# 2nd_name = "x"     # starts with a number
# my-name = "x"      # has a hyphen
```

### Data types
Every value in Python has a type. The four basic ones are:

| Type | Meaning | Example |
|------|---------|---------|
| `int` | whole number | `10` |
| `float` | decimal number | `3.14` |
| `str` | text | `"Python"` |
| `bool` | True or False | `True` |

Use `type()` to check the type of a value.

```python
whole_number = 10
decimal_number = 3.14
text = "Python"
is_learning = True

print(type(whole_number))
print(type(decimal_number))
print(type(text))
print(type(is_learning))
```

### Working with text (strings)
You can join strings with `+`, and an **f-string** (a string starting with `f`) lets you put variables inside `{}`.

```python
first = "Victor"
last = "Muchai"

full_name = first + " " + last      # joining strings
print(full_name)
print(f"Hello, {full_name}!")        # f-string
print(len(full_name))               # number of characters
```
