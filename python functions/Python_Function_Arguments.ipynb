# Python Functions
A **function** is a block of code with a name that you can run (call) whenever you need it. Functions help you avoid repeating code.

Simple function definition and calling

```python
def greet():
    print("Hello, Victor!")

greet()
```

A function is made of:
- the keyword `def`
- the function name
- brackets `()` (parameters go inside)
- a colon `:`
- an indented body

Once defined, a function can be called as many times as you like.

```python
greet()
greet()
greet()
```

Function with an argument

```python
def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Victor")
```

### Multiple parameters
A function can take more than one parameter. Separate them with commas.

```python
def add(a, b):
    print(a + b)

add(3, 5)
add(10, 20)
```

### Returning a value
`print()` only *shows* something on the screen. `return` sends a value **back** to the place where the function was called, so you can store it or use it again.

```python
def add_numbers(a, b):
    return a + b

result = add_numbers(3, 5)
print(result)
print(add_numbers(result, 10))
```

### Default parameter values
If a parameter has a default value, you don't have to provide it when calling the function.

```python
def greet_with_message(name, message="Welcome"):
    print(f"{message}, {name}!")

greet_with_message("Victor")
greet_with_message("Victor", "Good morning")
```

### Keyword arguments
You can name the arguments when you call the function. Then the order doesn't matter.

```python
def describe_pet(animal, name):
    print(f"{name} is a {animal}")

describe_pet("dog", "Rex")                 # positional: order matters
describe_pet(name="Tom", animal="cat")     # keyword: order does not matter
```

### Any number of arguments: `*args` and `**kwargs`
- `*args` collects extra positional arguments into a tuple.
- `**kwargs` collects extra keyword arguments into a dictionary.

```python
def add_all(*numbers):
    total = 0
    for n in numbers:
        total += n
    return total

print(add_all(1, 2, 3))
print(add_all(10, 20, 30, 40))
```

```python
def show_info(**details):
    for key, value in details.items():
        print(key, ":", value)

show_info(name="Victor", age=22)
```

### Returning more than one value
A function can return several values separated by commas. They come back as a tuple that you can unpack.

```python
def min_max(numbers):
    return min(numbers), max(numbers)

lowest, highest = min_max([4, 9, 1, 7])
print(lowest, highest)
```

### Docstrings
A string right under the `def` line describes what the function does. It is called a docstring.

```python
def square(n):
    """Return the square of n."""
    return n * n

print(square(4))
print(square.__doc__)
```

Local vs Global variable scope

```python
x = "global"  # Global variable

def my_func():
    x = "local"  # Local variable inside the function
    print(x)

my_func()
print(x)
```

Using the global keyword to modify a global variable inside a function

```python
x = 10

def update_value():
    global x
    x = 20

update_value()
print(x)  # Output will be 20
```

Using `global` works, but it is usually cleaner to pass values **in** as arguments and send results **out** with `return`:

```python
def add_ten(number):
    return number + 10

value = 5
value = add_ten(value)
print(value)
```

Importing a built-in module

```python
# Importing a built-in module
import math

print(math.sqrt(16))  # Calculates square root
```

More ways to use a module:

```python
import math

print(math.pi)
print(math.ceil(4.2))     # round up
print(math.floor(4.8))    # round down
print(math.pow(2, 3))     # 2 to the power of 3

from math import sqrt     # import just one function
print(sqrt(25))           # no need to write math. now
```

A function that calls itself (recursion)

```python
def countdown(n):
    if n <= 0:
        print("Done!")
    else:
        print(n)
        countdown(n - 1)

countdown(3)
```

A recursive function needs a **base case** (when to stop), otherwise it calls itself forever. Here the base case is `n == 1`.

```python
def factorial(n):
    if n == 1:                    # base case
        return 1
    return n * factorial(n - 1)   # the function calls itself

print(factorial(5))               # 5 * 4 * 3 * 2 * 1
```

Standard way to structure an executable script in Python

```python
def main():
    print("Program started successfully!")

if __name__ == "__main__":
    main()
```

`__name__` is a special variable. When you run a file directly, its value is `"__main__"`, so `main()` runs. When the file is imported into another file, its value is the module name, so `main()` does **not** run.

```python
print(__name__)
```
